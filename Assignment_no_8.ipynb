```python
#Dictionary
a={1:"pallavi",2:"dhanaraj",3:"sanskar",4:"pratiksha",5:"vinay"}
print(a)

#data type
print(type(a))

#length of dictionary
print(len(a))

#accessing the value from dictionary
print(a.values())

#accesing the keys from the dictionary
print(a.keys())

#accessing value using key
print(a[3])

#converting a dictionary in tuple/list

print(tuple(a))

print(list(a))
```
